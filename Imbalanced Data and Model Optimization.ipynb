## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, roc_auc_score
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
```

## 2 - Data Collection

```python
breast_cancer_data = datasets.load_breast_cancer()
breast_cancer_data
```

```python
breast_cancer_df = pd.DataFrame(data = breast_cancer_data.data, columns = breast_cancer_data.feature_names)
breast_cancer_df
```

```python
breast_cancer_df["Target"] = breast_cancer_data.target
breast_cancer_df
```

```python
breast_cancer_df["Target"].value_counts()
```

## 3 - Data Understanding

```python
breast_cancer_df.shape
```

```python
breast_cancer_df.isna().sum()
```

```python
breast_cancer_df.dtypes
```

## 4 - Data Preparation

```python
X = breast_cancer_df.drop(labels = "Target", axis = 1)
y = breast_cancer_df["Target"]
```

```python
X.shape,y.shape
```

```python
X_train, X_test, y_train, y_test  = train_test_split(X,y,test_size = 0.2, random_state = 123, stratify = y)
```

```python
plt.pie(breast_cancer_df["Target"].value_counts(), autopct = "%1.0f%%", labels = ["Malignant Cancer", "Benign Cancer"])
plt.show()
```

```python
plt.pie(y_train.value_counts(), autopct = "%1.0f%%",  labels = ["Malignant Cancer", "Benign Cancer"])
plt.show()
```

```python
plt.pie(y_test.value_counts(), autopct = "%1.0f%%",  labels = ["Malignant Cancer", "Benign Cancer"])
plt.show()
```

## 5 & 6 - Model Building and Model Training

```python
%%time
dt_model = DecisionTreeClassifier(criterion='gini', max_depth=3, class_weight={0:2,1:1})
dt_model.fit(X_train,y_train)
```

```python
%%time
ada_boost = AdaBoostClassifier(estimator=dt_model,n_estimators=25)
ada_boost.fit(X_train,y_train)
```

## 7 & 8 - Model Testing and Model Evaluation

### Decision Tree Model

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python
y_pred_test = dt_model.predict(X_test)
y_pred_test
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
fpr,tpr,thersholds = roc_curve(y_train, dt_model.predict_proba(X_train)[:,1])
auc = roc_auc_score(y_train,y_pred_train)
print(auc)

plt.plot(fpr,tpr,color = "red", label = "dt model (area =%0.1f)"%auc)
plt.plot([0,1],[0,1], "k--")
plt.xlabel("False Positive Rate or [1 - True Positive Rate]")
plt.ylabel("True Positive Rate")
plt.show()
```

```python
fpr,tpr,thersholds = roc_curve(y_test, dt_model.predict_proba(X_test)[:,1])
auc = roc_auc_score(y_test,y_pred_test)
print(auc)

plt.plot(fpr,tpr,color = "red", label = "dt model (area =%0.1f)"%auc)
plt.plot([0,1],[0,1], "k--")
plt.xlabel("False Positive Rate or [1 - True Positive Rate]")
plt.ylabel("True Positive Rate")
plt.show()
```
